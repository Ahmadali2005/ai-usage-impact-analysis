```python
import pandas as pd 

df=pd.read_csv(r"AI_Usage_and_Impact_on_Students_and_Professionals.csv")
print(df.info())
print(df.describe())
print(df.head())
```

```python
df.drop(columns=['User_ID'],inplace=True)
df.isna().sum()
```

```python
df['Gender']=df['Gender'].str.capitalize()
df['AI_Tool']=df['AI_Tool'].str.replace('Chat Gpt','ChatGpt').str.strip()
```

```python
df['AI_Usage_Hours_Per_Day'] = df['AI_Usage_Hours_Per_Day'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
df['Work_or_Study_Hours_Per_Day'] = df['Work_or_Study_Hours_Per_Day'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
df['Time_Saved_Hours_Per_Week'] = df['Time_Saved_Hours_Per_Week'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
df['Monthly_AI_Cost'] = df['Monthly_AI_Cost'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
df['Monthly_Income'] = df['Monthly_Income'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
```

```python
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
cat_cols = df.select_dtypes(include=['str']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df=df.drop_duplicates()

print(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.figure(figsize=(15,10))

plt.subplot(2,2,1)
sns.countplot(data=df,y='AI_Tool',order=df['AI_Tool'].value_counts().index,palette='viridis')
plt.title("1.most used ai tools")
plt.xlabel('Count')
plt.ylabel('AI Tool')

plt.subplot(2,2,2)
sns.boxplot(data=df,x='User_Type',y='AI_Usage_Hours_Per_Day',palette='Set2')
plt.title('2.Daily AI Usage Hours: Student vs Professional')
plt.xlabel('user type')
plt.ylabel('student vs professional')

plt.subplot(2, 2, 3)
sns.scatterplot(data=df, x='AI_Usage_Hours_Per_Day', y='Productivity_Score', hue='User_Type', alpha=0.7)
plt.title('3. Usage Hours vs Productivity Score')
plt.xlabel('Daily Usage Hours')
plt.ylabel('Productivity Score')

plt.subplot(2, 2, 4)
sns.countplot(data=df, x='Would_Recommend', hue='User_Type', palette='Set1')
plt.title('4. Recommendation Willingness by User Type')
plt.xlabel('Would Recommend?')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```
